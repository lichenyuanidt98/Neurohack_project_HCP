```python
import os
import pandas as pd
from pathlib import Path
from cloudpathlib import S3Path, S3Client
```

```python
local_dir = Path("./Neurohack_project_HCP/downloaded_connectomes")
local_dir.mkdir(parents=True, exist_ok=True)
```

```python
# S3 Client configuration
cache_path = Path('/tmp/cache')
cache_path.mkdir(parents=True, exist_ok=True)

client = S3Client(no_sign_request=True, local_cache_dir=cache_path)
root_path = S3Path('s3://open-neurodata/hcp1200/', client=client)
```

```python
space_folder = "Glasser_space-MNI152NLin6_res-1x1x1"
all_subjects = [p for p in root_path.iterdir() if p.is_dir() and p.name.startswith("sub-")]
target_subjects = all_subjects[:30]
print(f"Found {len(target_subjects)} subjects to process.")
```

```python
target_subjects
```

```python
for idx, sub_path in enumerate(target_subjects, start=1):
    sub_id = sub_path.name
    connectome_dir = sub_path / "ses-1" / "connectomes" / space_folder
    
    if not connectome_dir.exists():
        print(f"[{idx}/10] {sub_id}: Connectome folder not found. Skipping.")
        continue
        
    csv_files = list(connectome_dir.glob("*_connectome.csv"))
    if not csv_files:
        print(f"[{idx}/10] {sub_id}: CSV not found. Skipping.")
        continue
        
    s3_csv = csv_files[0]
    out_file = local_dir / f"{sub_id}_{space_folder}_connectome.csv"    
    s3_csv.download_to(out_file)
    print(f"[{idx}/10] Saved: {out_file}")
```
